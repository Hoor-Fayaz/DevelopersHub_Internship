```python
import yfinance as yf
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = yf.download("AAPL", start="2019-01-01", end="2026-01-01")
data.head()
```

```python
df = data.copy()

# Shift Close column upward by 1 to create "Next Day Close"
df["Target"] = df["Close"].shift(-1)

# Remove last row (no target available)
df.dropna(inplace=True)

# Select input features
features = ["Open", "High", "Low", "Volume"]
X = df[features]
y = df["Target"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, shuffle=False
)
```

```python
lr = LinearRegression()
lr.fit(X_train, y_train)
lr_pred = lr.predict(X_test)
```

```python
rf = RandomForestRegressor(n_estimators=200, random_state=42)
rf.fit(X_train, y_train)
rf_pred = rf.predict(X_test)
```

```python
mse = mean_squared_error(y_test, rf_pred)
print("Random Forest MSE:", mse)
```

```python
plt.figure(figsize=(12, 6))
plt.plot(y_test.values, label="Actual Close Price", linewidth=2)
plt.plot(rf_pred, label="Predicted Close Price", linewidth=2)
plt.title("Actual vs Predicted Closing Price")
plt.xlabel("Days")
plt.ylabel("Price ($)")
plt.legend()
plt.tight_layout
```
